```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Create the Dataset
data = {
    'House_Size': [1200, 1500, 800, 2000, 1100, 2200, 950],
    'Price': [200000, 350000, 100000, 500000, 220000, 550000, 150000],
    'Neighborhood': ['Suburb', 'City', 'Rural', 'City', 'Suburb', 'City', 'Rural']
}
df = pd.DataFrame(data)

# 2. Draw a Scatter Plot (Size vs Price)
plt.figure(figsize=(10, 5))
sns.scatterplot(data=df, x='House_Size', y='Price', hue='Neighborhood', s=100)

plt.title("House Prices vs Size")
plt.xlabel("Size (Sq Ft)")
plt.ylabel("Price ($)")
plt.grid(True)
plt.show()
```
